```r
# deconvolute the TIME for TCGA dataset

library(tibble)
library(immunedeconv)
```

```r
load("../data/prad_data_tcga.RData")
```

```r
# deconvolution
res.deconv <- deconvolute(as.matrix(prad.gene), "mcp_counter", tumor = TRUE)
```

```r
res.deconv
```

```r
# remove the row "cytotoxicity score" which is not a cell type
res.deconv <- res.deconv[res.deconv$cell_type != "cytotoxicity score", ]
```

```r
res.deconv
```

```r
res.deconv.mcpcounter <- column_to_rownames(res.deconv, colnames(res.deconv)[1])

res.deconv.sample <- as.data.frame(t(res.deconv.mcpcounter))
```

```r
time.mcpcounter.bcr <- merge(res.deconv.sample, df.clinical.bcr[, c("bcr_days", "biochemical_recurrence")], by = "row.names")
time.mcpcounter.bcr <- column_to_rownames(time.mcpcounter.bcr, colnames(time.mcpcounter.bcr)[1])
```

```r
write.csv(time.mcpcounter.bcr, "../data/time_mcpcounter_bcr_tcga.csv")
save(res.deconv.mcpcounter, prad.gene, df.clinical.bcr, time.mcpcounter.bcr, file = "../data/tcga_time_deconv_mcpcounter.RData")
```
