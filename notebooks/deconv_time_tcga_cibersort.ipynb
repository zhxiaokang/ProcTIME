```r
# deconvolute the TIME for TCGA dataset

library(tibble)
library(immunedeconv)
```

```r
load("../data/prad_data_tcga.RData")
```

```r
sig_matrix <- "../scripts/LM22.txt"
binary <- "../scripts/CIBERSORT.R"
set_cibersort_mat(sig_matrix)
set_cibersort_binary(binary)
```

```r
# deconvolution
res.deconv <- deconvolute(as.matrix(prad.gene), "cibersort_abs")
```

```r
res.deconv
```

```r
dim(res.deconv)
```

```r
res.deconv <- res.deconv[rowSums(res.deconv != 0) > 0, ]
dim(res.deconv)
```

```r
res.deconv.cibersort <- column_to_rownames(res.deconv, colnames(res.deconv)[1])

res.deconv.sample <- as.data.frame(t(res.deconv.cibersort))
```

```r
time.cibersort.bcr <- merge(res.deconv.sample, df.clinical.bcr[, c("bcr_days", "biochemical_recurrence")], by = "row.names")
time.cibersort.bcr <- column_to_rownames(time.cibersort.bcr, colnames(time.cibersort.bcr)[1])
```

```r
write.csv(time.cibersort.bcr, "../data/time_cibersort_bcr_tcga.csv")
save(res.deconv.cibersort, prad.gene, df.clinical.bcr, time.cibersort.bcr, file = "../data/tcga_time_deconv_cibersort.RData")
```
