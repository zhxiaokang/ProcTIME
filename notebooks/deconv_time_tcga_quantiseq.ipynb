```r
# deconvolute the TIME for TCGA dataset

library(tibble)
library(immunedeconv)
```

```r
load("../data/prad_data_tcga.RData")
```

```r
# deconvolution
res.deconv <- deconvolute(as.matrix(prad.gene), "quantiseq", tumor = TRUE)
```

```r
res.deconv.quantiseq <- column_to_rownames(res.deconv, colnames(res.deconv)[1])

res.deconv.sample <- as.data.frame(t(res.deconv.quantiseq))
```

```r
time.quantiseq.bcr <- merge(res.deconv.sample, df.clinical.bcr[, c("bcr_days", "biochemical_recurrence")], by = "row.names")
time.quantiseq.bcr <- column_to_rownames(time.quantiseq.bcr, colnames(time.quantiseq.bcr)[1])
```

```r
write.csv(time.quantiseq.bcr, "../data/time_quantiseq_bcr_tcga.csv")
```

```r
save(res.deconv.quantiseq, prad.gene, df.clinical.bcr, time.quantiseq.bcr, file = "../data/tcga_time_deconv_quantiseq.RData")
```
